# Features Description

Age → Age of the individual (years)

Gender → Gender of the individual

Height → Height of the individual (in meters)

Weight → Weight of the individual (in kilograms)

CALC → Consumption of Alcohol

FAVC → Frequent consumption of High Caloric food

FCVC → Frequency of Consumption of Vegetables

NCP → Number of main meals (per day)

SCC → Calories Consumption Monitoring

SMOKE → Smoking habit

CH2O → Consumption of Water (daily water intake)

family_history_with_overweight → Family history with overweight/obesity

FAF → Frequency of Physical Activity

TUE → Time Using Technology devices

CAEC → Consumption of food Between meals

MTRANS → Mode of Transportation

NObeyesdad → Nutritional Obesity Level (Obesity diagnosis)

## **Data Cleaning**

**Libraries**

```python
import pandas as pd
```

```python
df =pd.read_csv('/content/Updated_Obesity_Dataset.csv')
```

```python
df.shape
df.columns
```

```python
df.head()
```

```python
df.info()
```

```python
df['NObeyesdad'].value_counts()
```

Null Vals

```python
df.isnull().sum()
```

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
```

```python
df['CH2O'].fillna(df['CH2O'].median(), inplace=True)
```

```python
df['family_history_with_overweight'].fillna(df['family_history_with_overweight'].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

Duplicates

```python
df.duplicated().sum()
```

```python
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows
```

```python
df.drop_duplicates(inplace=True)
```

Scaling & Encoding

```python
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns
print(numerical_features)
```

```python
categorical_features = df.select_dtypes(include=['object']).columns
categorical_features
```

Train-test splitting

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
x=df.drop('NObeyesdad',axis=1)
y=df['NObeyesdad']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=123)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
categorical_features=x.select_dtypes(include=['object']).columns
numerical_features=x.select_dtypes(include=['int64', 'float64']).columns
```

```python
scaler=StandardScaler()

x_train[numerical_features] = scaler.fit_transform(x_train[numerical_features])
x_test[numerical_features]= scaler.transform(x_test[numerical_features])
```

```python
print('Head of x_train after scaling:')
display(x_train[numerical_features].head())
```

```python
print('\nMean of numerical features in x_train after scaling:')
display(x_train[numerical_features].mean())
print('\nStandard deviation of numerical features in x_train after scaling:')
display(x_train[numerical_features].std())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in categorical_features:
    x_train[col] = le.fit_transform(x_train[col])
    x_test[col] = le.transform(x_test[col])
```

```python
print('x_train after all transformations:')
display(x_train.head())
print('\nx_test after all transformations:')
display(x_test.head())
```

# **NAIVE BAYES**

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
model.fit(x_train,y_train)
```

```python
nb_y_pred = model.predict(x_test)
#print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, nb_y_pred))
print(classification_report(y_test, nb_y_pred))
print(confusion_matrix(y_test, nb_y_pred))
```

# **KNN**

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(x_train, y_train)
```

```python
knn_y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, knn_y_pred))
print(classification_report(y_test, knn_y_pred))
print(confusion_matrix(y_test, knn_y_pred))
```
